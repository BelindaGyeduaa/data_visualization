```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
x = np.arange(1, 12, 2)
y = np.arange(1, 12, 2)

# Creat a line chart with dashdot linestyle
plt.plot(x, y, linestyle='dashdot', color='#F5B027', marker='*', markersize=20)
plt.show()
```

```python
fig, ax = plt.subplots()

# Creat a line chart with dash linestyle
plt.plot(x, y, linestyle='--', color='green', marker='v')
plt.show()
```

```python
fig, ax = plt.subplots()

# Creat a line chart with dotted linestyle
plt.plot(x, y, linestyle='dotted', color='red', marker='v', markersize=10)
plt.show()
```

```python
x = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
# Create a dashed line with diamond markers
plt.plot(y, linestyle='--', color='red', marker='D')
plt.show()
```

```python
# Create the opposite approach, linestyle of dashdot, and squre markers
plt.plot(y, linestyle='dashdot', color='violet', marker='s')
plt.show()
```

```python
# diamond marker, red color marker, marker width
plt.plot(y, linestyle='dashdot', color='black', marker='D', markersize=10, markerfacecolor='red', markeredgewidth=2)
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
# Create a lineplot, custom color, dotted linestyle, diamond markers
fig, ax = plt.subplots()

# Creat a line chart with dotted linestyle
# Create a lineplot, custom color, dotted line
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color=(0.5, 0.5, 0.8)
)
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.show()
```

```python
# Create a lineplot, custom color, dotted line
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color='#ff0040'
)
plt.title('Sports Watch Data', color='blue', style='italic', fontweight='bold')
plt.xlabel('Average Pulse', color='orange')
plt.ylabel('Calorie Burnage', color='green')
plt.show()
```

```python
years=[1970,1980,1985,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops=[3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.3,7.4,7.5]
```

```python
# title of 'World Population' - blue
# labels growth by year, population in billions green red
fig, ax = plt.subplots()
plt.plot(
    years, pops,
    linestyle=':',
    marker='v',
    color='#bf00ff',
     markerfacecolor='blue',
     markersize=8
)
plt.title('World Population', color='blue', style='italic', fontweight='bold')
plt.xlabel('Growth by year', color='green')
plt.ylabel('Population (billions)', color='red')
plt.show()
```

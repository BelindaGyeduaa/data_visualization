### Matplotlib Functions
- `plt.plot()` - defualt creates line chart (trends over time)
- `plt.show()` - start an eventloop that opens interactive window to view active figures (fig), similar to print() but for plots
- `plt.scatter()` - scatter plot, visualizing relationships
- `plt.bar()` - comparing categorical or discrete data.
- `plt.barh()` - hz barplot
- `plt.hist()` - barplot for continuous data/quantitative data
- `plt.savefig()` - can save a plot to a file
- `plt.contour()` - draws contour lines
- `plt.annotate()` - create annotation on your plot

#### Plot Vocabulary:

- figure (fig) - container, window to put your plots on
    - figsize - size of the figure in inches
    - dpi - dots per inch (resolution)
    - facecolor - background color
    - subplotspars - distance between subplots
- axes (ax) - area within the fig where data is plotted
    - title - title for the chart/plot
    - xlabel, ylabel, xscale, yscale, xlim, ylim
- axis - x, y axis of a plot or chart
- line - data itself, cant be fully customized
    - color
    - linestyle
    - linewidth
    - markers - symbols that represent data points
    - markersize
- Legend - small container with labels/descriptors of elements
- grid - backgrod (attribute that can be edited)
- ticks (tic) - marks along the x,y axis to represent values

### Approaches to Creating Plots:

```python
# state based approach - implicitly create and manipulate figures. Less upfront code, but less control
x_points = np.array([1, 13])
y_points = np.array([2, 12])
plt.plot(x_points, y_points, color='green')
plt.show()
```

```python
# object oriented approach - explicitly creating and manipulating your axes and figures. More upfront code, but more control
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x_points, y_points,  linestyle='--', color='#F5B027')
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
plt.plot(
    x, y,
    linestyle='--',
    marker='D',
    color=(0.5, 0.5, 0.8)
)
plt.title('Sports Watch Data', color='blue', style='italic', fontweight='bold')
plt.xlabel('Average Pulse', color='green')
plt.ylabel('Calorie Burnage', color='green')
plt.grid(color='red', linestyle=':', alpha=0.2)
plt.gca().set_facecolor('lightyellow') # inside the graph
plt.gcf().set_facecolor('lightblue') # outside the graph
plt.show()
```

```python
plt.figure(figsize=(3, 3), dpi=144) # helps adjust the size 

plt.plot(
    x, y,
    linestyle='--',
    marker='D',
    color=(0.5, 0.5, 0.8)
)
plt.title('Sports Watch Data', color='blue', style='italic', fontweight='bold')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.grid(color='red', linestyle=':', alpha=0.2)
plt.gca().set_facecolor('lightyellow') # inside the graph
plt.gcf().set_facecolor('lightblue') # outside the graph
plt.show()
```

```python
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)  # create 1 row, with 2 plots, this plot is index 1
plt.plot(x, y)
plt.title('Plot 1')

x1 = np.array([0, 1, 2, 3])
y1 = np.array([10, 20, 25, 30])

plt.subplot(1, 2, 2)  # create 1 row, with 2 plots, this plot is index 2
plt.title('Plot 2')
plt.plot(x1, y1)
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 3, 5, 7, 11])
y2 = np.array([1, 4, 9, 16, 25])

fig, axs = plt.subplots(2, 1)

axs[0].plot(x, y1)
axs[0].set_title('Plot 1')

axs[1].plot(x, y2)
axs[1].set_title('Plot 2')

plt.tight_layout()  # spaces plots apart to avoid overlapping
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x)

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sin Function')

axes[0, 1].plot(x, y2)
axes[0, 1].set_title('Cosine Function')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Tangent Function')

axes[1, 1].plot(x, y4)
axes[1, 1].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 1, sharex=True)

axes[0].plot(x, y1)
axes[0].set_title('Sin Function')

axes[1].plot(x, y2)
axes[1].set_title('Cosine Function')

axes[2].plot(x, y3)
axes[2].set_title('Tangent Function')

axes[3].plot(x, y4)
axes[3].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
years = [1950, 1955, 1960, 1965, 1970, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015]
pops = [2.5, 2.7, 3.0, 3.3, 3.6, 4.0, 4.4, 4.8, 5.3, 5.7, 6.1, 6.5, 7.3]
death = [1.2, 1.1, 1.2, 1.2, 1.0, 2.3, 1.8, 1.9, 2.6, 1.6, 2.4, 2.4, 4.0]
job_rate = [0.5, 0.6, 0.8, 0.9, 1.0, 1.2, 1.5, 1.7, 2.0, 2.2, 2.5, 2.7, 2.9]

plt.plot(years, pops, color='red', linestyle='--', label='Population', marker='D', markerfacecolor='blue')
plt.plot(years, death, color='green', label='Death')
plt.plot(years, job_rate, color='purple', linestyle='-.', label='Job Rate', marker='v')

plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population (Billions)')

plt.legend()  # when using a legend, the plots must have labels
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```

```python
# Object Oriented (OO)
fig, axes = plt.subplots()

axes.plot(years, pops, color='red', linestyle='--', label='Population')
axes.plot(years, death, color='green', linestyle=':', label='death')

plt.legend() # When using a legend, the plot must have a label. See .polt() above

plt.show()
```
